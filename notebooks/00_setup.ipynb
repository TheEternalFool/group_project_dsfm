# Check the working environment

Run this notebook with the project `.venv` interpreter. It checks Python, pandas and relative paths only; it does not read or analyse the SSA data.

Launch JupyterLab from the repository root, then open this notebook. The path check also supports running it with the root as the working directory.


```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

assert (PROJECT_ROOT / "requirements.txt").is_file(), "Open JupyterLab from the repository root."
DATA_DIR = PROJECT_ROOT / "data"
RAW_STATE_DIR = DATA_DIR / "raw" / "state"
RAW_NATIONAL_DIR = DATA_DIR / "raw" / "national"
PROCESSED_DIR = DATA_DIR / "processed"

for directory in (RAW_STATE_DIR, RAW_NATIONAL_DIR, PROCESSED_DIR):
    assert directory.is_dir(), f"Missing project directory: {directory}"

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("pandas:", pd.__version__)
print("Project root:", PROJECT_ROOT)
print("Original state data:", RAW_STATE_DIR)
print("Original national data:", RAW_NATIONAL_DIR)
print("Prepared data:", PROCESSED_DIR)
```

The reference environment uses Python 3.12; the interpreter should belong to this project's `.venv`.
These directory checks also pass in a fresh clone without any downloaded data. They do not verify data availability or quality. See `data/README.md` for source files and their current status.
